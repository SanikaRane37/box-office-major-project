```python
from sklearn.feature_extraction.text import CountVectorizer
cv_trial=CountVectorizer()
X=cv_trial.fit_transform(df['all_tags']).toarray()
print(X)
```

```python
y=df.iloc[:,-1].values
print(y.shape)
y
```

```python

X
```

```python
from collections import Counter
from imblearn.over_sampling import SMOTE
sm = SMOTE()
print('UNBALANCED_DATA',Counter(y))
x_sm,y_sm = sm.fit_resample(X,y)
print('BALANCED_DATA',Counter(y_sm))
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(x_sm,y_sm,test_size=0.2,random_state=10)

print(X_train.shape,X_test.shape,y_train.shape,y_test.shape)
```

```python
from sklearn.naive_bayes import GaussianNB,MultinomialNB,BernoulliNB
from sklearn.metrics import accuracy_score
```

```python
clf2=MultinomialNB()

clf2.fit(X_train,y_train)

y_pred2 = clf2.predict(X_test)

print('MultinomialNB', accuracy_score(y_test, y_pred2))
```

```python
clf1=GaussianNB()

clf1.fit(X_train,y_train)

y_pred1 = clf1.predict(X_test)

print('Gaussian', accuracy_score(y_test, y_pred1))
```

```python
clf3=BernoulliNB()

clf3.fit(X_train,y_train)

y_pred3 = clf3.predict(X_test)

accuracy=accuracy_score(y_test, y_pred3)

print('BernoulliNB', round(accuracy,2))


```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(n_estimators=400)

# Train the classifier on the training data
rf.fit(X_train, y_train)

# Make predictions on the test data
predictions = rf.predict(X_test)

# Calculate the accuracy
accuracy = accuracy_score(y_test, predictions)

print("Accuracy:", accuracy)
```
