```python
def remove_special_char(text):
    x=''
    
    for i in text:
        if i.isalnum():
            x = x + i
        else:
            x = x + ' '
    return x
```

```python
def convert_lower(text):
    l=[]
    for item in text:
        l.append(item.lower())
    return l
```

```python
new_df['all_tags']=new_df['all_tags'].apply(convert_lower)
```

```python
import nltk
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
def steming(text):
    l=[]
    for i in text:

        l.append(ps.stem(i))
    return l
```

```python
new_df['all_tags']=new_df['all_tags'].apply(steming)
```

```python
def _join(x):
    l=[]
    for i in x:
        l.append(i.replace(" ",""))
    return l
```

```python
# VECTORIZATION

from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(stop_words='english',max_features=5000)
```

```python
data = cv.fit_transform(new_df['all_tags']).toarray()
data
```

```python
data.shape
```

```python
# COSINE SIMILARITY

from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(data)
similarity
```

```python
import pickle

pickle.dump(new_df,open("movie_recm1.pkl","wb"))
pickle.dump(similarity,open("similarity1.pkl","wb"))
```

```python
def movie_recommend(movie,number=3):
    index = new_df[new_df['Title'] == movie].index[0]
    distances = sorted(list(enumerate(similarity[index])),reverse=True,key = lambda x: x[1])
    for i in distances[1:number+1]:
        print("Title: {}".format(new_df.iloc[i[0]].Title))
        print("tagline: {}".format(new_df.iloc[i[0]].tagline))
        print("Overview: {}".format(new_df.iloc[i[0]].overview))
        print("Cast: {}".format(new_df.iloc[i[0]].full_cast))
        print("Genres: {}".format(new_df.iloc[i[0]].genres))
        print("Link: {}".format(new_df.iloc[i[0]].Title_URL))
        print("  ")
        print("  ")
```

```python
movie_recommend("Harry Potter and the Goblet of Fire",5)
```
